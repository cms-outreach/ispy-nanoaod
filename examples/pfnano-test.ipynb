# PFNano in ispynanoaod

PFNano is an enhanced NANOAOD format 

We'll look at this dataset, [MuOnia dataset in NanoAOD format enhanced with Particle Flow candidates from RunG of 2016 ](https://opendata.cern.ch/record/31310).

In addition to the default NanoAOD content, it contains candidates from the Particle Flow algorithm. Their properties are available in the "PFCands" collection.

You can learn more [here](https://cms-opendata-workshop.github.io/workshop2023-lesson-advobjects/02-particleflow/index.html) about what Particle Flow is. 

Here's what is in the PFCands collection in PFNano:
